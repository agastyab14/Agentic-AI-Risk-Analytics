```python
import os
from dotenv import load_dotenv
from fredapi import Fred
```

```python
load_dotenv()
fred = Fred(api_key=os.getenv("FRED_API_KEY"))
```

```python
cpi = fred.get_series("INDCPIALLMINMEI")
print(cpi.tail())
```

```python
latest_value = cpi.iloc[-1]
prior_value = cpi.iloc[-2]

print("Latest:", latest_value)
print("Prior:", prior_value)
```

```python
pct_change = (latest_value - prior_value) / prior_value
print("Percent change:", pct_change)
```

```python
threshold = 0.0025  # 0.25%, as per your build guide's example

is_surprise = abs(pct_change) > threshold
print("Is this a surprise?", is_surprise)
```

```python
def check_macro_surprise(series_id="INDCPIALLMINMEI", threshold=0.0025):
    series = fred.get_series(series_id)
    latest_value = series.iloc[-1]
    prior_value = series.iloc[-2]
    pct_change = (latest_value - prior_value) / prior_value

    if abs(pct_change) > threshold:
        return {
            "series_id": series_id,
            "latest_value": latest_value,
            "prior_value": prior_value,
            "pct_change": pct_change,
            "surprise": True
        }
    else:
        return None
```

```python
result = check_macro_surprise()
print(result)
```
